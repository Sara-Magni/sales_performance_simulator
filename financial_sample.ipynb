```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv("data/data_raw/financial_sample_raw.csv")
```

```python
data.shape
```

```python
data.columns
```

```python
# Rimuovo gli spazi vuoti prima e dopo i nomi di tutte le colonne
data.columns = data.columns.str.strip()
```

```python
data.columns
```

```python
# Normalizzo i nomi delle colonne
data.columns = data.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
data.head(20)
```

```python
data.duplicated().sum()
```

```python
data.info()
```

```python
# Conversione delle colonne monetarie in float
currency_cols = ['manufacturing_price', 'sale_price', 'gross_sales', 'discounts', 'sales', 'profit']

for col in currency_cols:
    if col in data.columns:
        data[col] = (
            data[col].astype(str)
            .str.replace('$', '', regex=False)
            .str.replace(',', '', regex=False)
            .str.replace('(', '-', regex=False)  # Converte la parentesi aperta in meno
            .str.replace(')', '', regex=False)   # Rimuove la parentesi chiusa
            .str.replace('-', '0', regex=False)
            .str.strip()
        )
        data[col] = pd.to_numeric(data[col], errors='coerce')

```

```python
data.head()
```

```python
# Converto la colonna Data in datetime
data['date'] = pd.to_datetime(data['date'])
```

```python
data.head()
```

```python
data.info()
```

```python
# Calcolo la colonna COGS
data['cogs'] = data['sales'] - data['profit']
```

```python
# Reset dell'indice
data = data.reset_index(drop=True)
```

```python
# Esportazione in CSV senza includere l'indice di pandas
data.to_csv('financial_data.csv', index=False)
```
